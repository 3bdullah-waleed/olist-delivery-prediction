```python
import pandas as pd

df = pd.read_csv("ml_table_step1.csv", parse_dates=[
    "order_purchase_timestamp", "order_delivered_customer_date", "order_estimated_delivery_date"
])

# Drop orders that were never delivered (target undefined for them)
df = df.dropna(subset=["order_delivered_customer_date"]).copy()

# Label: 1 = late, 0 = on time
df["is_late"] = (df["order_delivered_customer_date"] > df["order_estimated_delivery_date"]).astype(int)

print(df["is_late"].value_counts(normalize=True))

df.to_csv("ml_table_labeled.csv", index=False)
print("Saved:", df.shape)
```
